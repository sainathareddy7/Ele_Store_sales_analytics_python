```python
#12) Query the total sales by region and country?
#c_data.head(5)
total_sales = c_data.groupby(['Region','Country'])['Sales'].sum().reset_index()
total_sales.rename(columns={'Sales': 'Total_Sales'},inplace=True)
print(total_sales)
```

```python
#18) Query the total order qty by Manufacturer and Product Name?
total_order_qty = c_data.groupby(['Manufacturer', 'ProductName'])['OrderQty'].sum().reset_index()
total_order_qty.rename(columns={'OrderQty':'Total_Qty'},inplace=True)
print(total_order_qty)
```

```python
#19) Query the top 7 products with orderqty and sales and show it descending order based on total orderqty?
top_7_products = c_data.groupby('ProductName')[['OrderQty', 'Sales']].sum().reset_index()

top_7_products = top_7_products.sort_values(by='OrderQty', ascending=False).head(7)

print(top_7_products)
```

```python
#20) Query the total order quantity greater than 10000 by city?
city_order_qty = c_data.groupby('City')['OrderQty'].sum().reset_index()

high_order_cities = city_order_qty[city_order_qty['OrderQty'] > 10000]

print(high_order_cities)
```

```python
#21) Query the total sales less than 1000 by product, display it in ascending order?
product_total_sales= c_data.groupby('ProductName')['Sales'].sum().reset_index()
less_sales_products= product_total_sales[product_total_sales['Sales']<1000]
less_sales_products_sortby = less_sales_products.sort_values(by='Sales', ascending=True)

print(less_sales_products_sortby)
```

```python
#22) Query the total profit greater than 200000 by product, display it in descending order?

high_profit_products = c_data.groupby('ProductName')['Profit'].sum().reset_index()
total_profit_products= high_profit_products[high_profit_products['Profit']>200000]
total_profit_products_sort=total_profit_products.sort_values(by='Profit',ascending =False)
print(total_profit_products_sort)
```
